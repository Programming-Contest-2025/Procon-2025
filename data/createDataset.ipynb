### 2. Tạo dữ liệu huấn luyện

```python
from utils import generateField
from Solver import Solver, build_mapping
from FieldClass import Field
import pandas as pd

sizes = range(4, 25, 2)
df = {
    "State": [],
    "FinalScore": [],
}

#Khởi tạo siêu tham số
Tstart_list = [1, 2, 3, 4, 5]


for size in sizes:
    mapping = build_mapping(n = size)
    for i in range(size * 2):
        entities = generateField(size)
        
        #Lấy điểm số cuối cùng của state trên
        custom_field = Field(size = size, entities=entities, mappings=mapping)
        solver = Solver(init_field = custom_field)
        
        count = 0
        for step, (state, score) in enumerate(solver.simulated_annealing_record(max_neiborghs=15 * size, alpha=0.995)):
            if step % 20 == 0:
                df["State"].append(state)
                df["FinalScore"].append(score)
                count += 1
                #Một path chỉ lấy 1000 (state, score)
                if count == 1000:
                    break
        print(f"Đã tạo bộ dữ liệu thứ {i + 1} cho size = {size}")

df = pd.DataFrame(df)
df.to_csv("./data/trainSet.csv", index=False, encoding='utf-8-sig')
print(f"TẠO DỮ LIỆU NHÂN TẠO THÀNH CÔNG !")
df
```

## 3. Tiền xử lý

```python
import pandas as pd
import os 
# Số lượng record mỗi size
df = pd.read_csv("trainSet.csv")
df.info()
```

### 3.1 Lấy max finalScore đối với các state trùng nhau

```python
new_df = df.groupby(by = "State", as_index=False)["FinalScore"].max()
new_df
```

### 3.2 Thêm cột size để train model theo từng size

```python
import ast
import numpy as np
new_df['Size'] = new_df['State'].apply(lambda x : 
    len(np.array(ast.literal_eval(x), np.int16)))
```

### 3.3 Chuyển các giá trị cột state thành list

```python
new_df['State'] = new_df['State'].apply(lambda x : ast.literal_eval(x))
new_df
```
